```python
import pandas as pd
```

```python
df = pd.read_csv('car_purchasing.csv',encoding="latin1")
```

```python
df.shape
```

```python
df.head()
```

```python
df = df.drop(['customer name','customer e-mail'],axis=1)
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df['country'].unique()
```

```python
counts = df['country'].value_counts()
df['country'] = df['country'].where(df['country'].map(counts) >=2, 'other')
```

```python
df['country'].nunique()
```

```python
X = df.drop('car purchase amount',axis=1)
y = df['car purchase amount']
```

```python
from sklearn.model_selection import  train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=42, random_state=42)
```

df.info()

```python
df.info()
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

cat_col = ['country']
scale_col = ['age', 'annual Salary', 'credit card debt','net worth']

preprocessor = ColumnTransformer(transformers=[
    ('ss', StandardScaler(), scale_col),
    ('encode', OneHotEncoder(handle_unknown='ignore', drop='first',sparse_output=False), cat_col)
], remainder='passthrough')
```

```python
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
```

```python
X_train.shape
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred = lr.predict(X_test)
y_pred_train = lr.predict(X_train)

print('train')

print(root_mean_squared_error(y_train, y_pred_train))
print(mean_squared_error(y_train, y_pred_train))
print(mean_absolute_error(y_train, y_pred_train))
print(r2_score(y_train, y_pred_train))

print('--test--')
print(root_mean_squared_error(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
print(r2_score(y_test, y_pred))
```

```python
from sklearn.linear_model import Lasso


lasso = Lasso(alpha=1.5)
lasso.fit(X_train, y_train)

y_pred = lasso.predict(X_test)
y_pred_train = lasso.predict(X_train)

print('train')

print(root_mean_squared_error(y_train, y_pred_train))
print(mean_squared_error(y_train, y_pred_train))
print(mean_absolute_error(y_train, y_pred_train))
print(r2_score(y_train, y_pred_train))

print('--test--')
print(root_mean_squared_error(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
print(r2_score(y_test, y_pred))
```

```python
from sklearn.linear_model import Ridge


ridge = Ridge(alpha=1.5)
ridge.fit(X_train, y_train)

y_pred = ridge.predict(X_test)
y_pred_train = ridge.predict(X_train)

print('train')

print(root_mean_squared_error(y_train, y_pred_train))
print(mean_squared_error(y_train, y_pred_train))
print(mean_absolute_error(y_train, y_pred_train))
print(r2_score(y_train, y_pred_train))

print('--test--')
print(root_mean_squared_error(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
print(r2_score(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestRegressor


rf = RandomForestRegressor(n_estimators=101,max_depth=3)
rf.fit(X_train, y_train)

y_pred = rf.predict(X_test)
y_pred_train = rf.predict(X_train)

print('train')

print(root_mean_squared_error(y_train, y_pred_train))
print(mean_squared_error(y_train, y_pred_train))
print(mean_absolute_error(y_train, y_pred_train))
print(r2_score(y_train, y_pred_train))

print('--test--')
print(root_mean_squared_error(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
print(r2_score(y_test, y_pred))
```

```python
from sklearn.neighbors import KNeighborsRegressor


knn = KNeighborsRegressor(n_neighbors=2)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
y_pred_train = knn.predict(X_train)

print('train')

print(root_mean_squared_error(y_train, y_pred_train))
print(mean_squared_error(y_train, y_pred_train))
print(mean_absolute_error(y_train, y_pred_train))
print(r2_score(y_train, y_pred_train))

print('--test--')
print(root_mean_squared_error(y_test, y_pred))
print(mean_squared_error(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
print(r2_score(y_test, y_pred))
```
