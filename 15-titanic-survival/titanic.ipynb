```python
import pandas as pd
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df.shape
```

```python
df = df.drop_duplicates()
df = df.drop(['PassengerId','Cabin','Name',],axis=1)
```

```python
df.head()
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df.isnull().sum()
```

```python
df['Ticket-category'] = df['Ticket'].str.replace(r'\d', '', regex=True)
df['Ticket-category'] = df['Ticket-category'].str.replace(r'[\s./]+', '', regex=True)
df['Ticket-category'] = df['Ticket-category'].replace('', 'NONE')
```
