```python
target = "T (degC)"

# Create a clean hourly target series
ts = hourly_df[target].copy()

print("Target:", target)
print("Total observations:", len(ts))

print("\nMissing values:")
print(ts.isnull().sum())

print("\nTemperature statistics:")
display(ts.describe())
```

```python
data = hourly_df.copy()

target = "T (degC)"


data["lag_1"] = data[target].shift(1)

data["lag_3"] = data[target].shift(3)

data["lag_6"] = data[target].shift(6)

data["lag_24"] = data[target].shift(24)

data["lag_168"] = data[target].shift(168)



data["rolling_mean_3"] = (
    data[target]
    .shift(1)
    .rolling(window=3)
    .mean()
)

data["rolling_mean_24"] = (
    data[target]
    .shift(1)
    .rolling(window=24)
    .mean()
)


data["hour"] = data.index.hour

data["day_of_week"] = data.index.dayofweek

data["month"] = data.index.month

data["day_of_year"] = data.index.dayofyear



data = data.dropna()


print("Feature engineered dataset shape:", data.shape)

display(data.head())

print("\nFeature columns:")
print(data.columns.tolist())
```

```python
features = [
    "lag_1",
    "lag_3",
    "lag_6",
    "lag_24",
    "lag_168",
    "rolling_mean_3",
    "rolling_mean_24",
    "hour",
    "day_of_week",
    "month",
    "day_of_year"
]

X = data[features]

y = data[target]
```

```python
split_index = int(len(data)*0.8)
X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred_baseline = X_test['lag_1']

mae = mean_absolute_error(
    y_test,
    y_pred_baseline
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_baseline
    )
)

r2 = r2_score(
    y_test,
    y_pred_baseline
)
print(f"MAE:  {mae:.4f} °C")
print(f"RMSE: {rmse:.4f} °C")
print(f"R²:   {r2:.4f}")
```

```python
results = pd.DataFrame({
    "Actual": y_test,
    "Predicted": y_pred_baseline
})

results["Error"] = (
    results["Actual"] - results["Predicted"]
)

display(results.head(10))
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
X_train_scale = ss.fit_transform(X_train)
X_test_scale = ss.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

rf.fit(
    X_train,
    y_train
)

y_pred = rf.predict(X_test)
y_pred_train = rf.predict(X_train)

print('train')

print(mean_absolute_error(y_train, y_pred_train))
print(np.sqrt(mean_squared_error(y_train, y_pred_train)))
print(r2_score(y_train, y_pred_train))

mae_rf = mean_absolute_error(
    y_test,
    y_pred
)

rmse_rf = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

r2_rf = r2_score(
    y_test,
    y_pred
)


print("\nRANDOM FOREST RESULTS")
print("-" * 35)

print(f"MAE:  {mae_rf:.4f} °C")
print(f"RMSE: {rmse_rf:.4f} °C")
print(f"R²:   {r2_rf:.4f}")
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(
    X_train_scale,
    y_train
)

y_pred = lr.predict(X_test_scale)
y_pred_train = lr.predict(X_train_scale)

print('train')

print(mean_absolute_error(y_train, y_pred_train))
print(np.sqrt(mean_squared_error(y_train, y_pred_train)))
print(r2_score(y_train, y_pred_train))

print('test')
print(mean_absolute_error(y_test, y_pred))
print(np.sqrt(mean_squared_error(y_test,y_pred)))
print(r2_score(y_test,y_pred))

```

```python
from sklearn.linear_model import Lasso

lasso = Lasso()

lasso.fit(
    X_train_scale,
    y_train
)

y_pred = lasso.predict(X_test_scale)
y_pred_train = lasso.predict(X_train_scale)

print('train')

print(mean_absolute_error(y_train, y_pred_train))
print(np.sqrt(mean_squared_error(y_train, y_pred_train)))
print(r2_score(y_train, y_pred_train))

print('test')
print(mean_absolute_error(y_test, y_pred))
print(np.sqrt(mean_squared_error(y_test,y_pred)))
print(r2_score(y_test,y_pred))

```
