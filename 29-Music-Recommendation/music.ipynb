```python
import ast

df['artists'] = df['artists'].apply(
    lambda x: ', '.join(ast.literal_eval(x))
    if isinstance(x, str) and x.startswith('[')
    else str(x)
)
```

```python
df.isnull().sum()
```

```python
features = [
    'valence',
    'acousticness',
    'danceability',
    'energy',
    'instrumentalness',
    'liveness',
    'loudness',
    'speechiness',
    'tempo',
    'key',
    'mode'
]
```

```python
df = df.dropna(subset=features)

df = df.drop_duplicates(subset=['name', 'artists'])
```

```python
df.head()
```

```python
df.info()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = scaler.fit_transform(df[features])
```

```python
X.shape
```

```python
indices = pd.Series(df.index, index=df['name']).drop_duplicates()
```

```python
from sklearn.neighbors import NearestNeighbors

model = NearestNeighbors(
    metric='cosine',
    algorithm='brute'
)
model.fit(X)
```

```python
def recommendation(song_name, n = 5):

    idx = indices[song_name]

    distance, indexes = model.kneighbors(
        X[idx].reshape(1,-1), n_neighbors=n+1
    )

    recommended_idx = indexes[0][1:]

    return df.iloc[recommended_idx][
        ['name', 'artists', 'year', 'popularity']
        ]
    
```

```python
recommendation("Havana", 5)
```
