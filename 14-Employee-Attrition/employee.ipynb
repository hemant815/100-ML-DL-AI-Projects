```python
import pandas as pd
```

```python
df =pd.read_csv('employee_attrition_dataset.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df = df.drop('Employee_ID',axis=1)
```

```python
df.describe()
```

```python
df.head()
```

```python
df.info()
```

```python
df['Overtime'] = df['Overtime'].map({'Yes':1, 'No':0})
df['Attrition'] = df['Attrition'].map({'Yes':1, 'No':0})
```

```python
X = df.drop('Attrition', axis=1)
y = df['Attrition']
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
df.iloc[:,20:]
```

```python
X_train.shape
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

scale_col = ['Age','Monthly_Income','Hourly_Rate','Years_at_Company','Years_in_Current_Role','Training_Hours_Last_Year','Average_Hours_Worked_Per_Week','Absenteeism','Distance_From_Home']
encode_col = ['Gender', 'Marital_Status', 'Department', 'Job_Role']

preprocessor = ColumnTransformer(transformers=[('ss', StandardScaler(), scale_col),
                                               ('cat', OneHotEncoder(handle_unknown='ignore',sparse_output=False, drop='first'), encode_col),
                                        
],remainder='passthrough')
```

```python
X_train_preproced = preprocessor.fit_transform(X_train)
X_test_preproced = preprocessor.transform(X_test)
```

```python
X_train_preproced.shape
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train_preproced, y_train)
```

```python
y_train_smote.value_counts()
```

```python
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, classification_report, f1_score, recall_score
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=1000, random_state=42)
lr.fit(X_train_smote, y_train_smote)
y_pred = lr.predict(X_test_preproced)
y_pred_train = lr.predict(X_train_smote)

print('train')
print(accuracy_score(y_train_smote, y_pred_train))
print(precision_score(y_train_smote, y_pred_train))
print(recall_score(y_train_smote, y_pred_train))
print(f1_score(y_train_smote, y_pred_train))
print(confusion_matrix(y_train_smote, y_pred_train))
print(classification_report(y_train_smote, y_pred_train))

print("-------test ----")
print(accuracy_score(y_test, y_pred))
print(precision_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(f1_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

result =[]
result.append({
    'model':'logitic',
    'train accuracy':accuracy_score(y_train_smote, y_pred_train),
    'test accuracy':accuracy_score(y_test, y_pred),
    'train precision': precision_score(y_train_smote, y_pred_train),
    'test precision': precision_score(y_test, y_pred),
    'train recall': recall_score(y_train_smote, y_pred_train),
    'test recall': recall_score(y_test, y_pred),
    'train f1 ':f1_score(y_train_smote, y_pred_train),
    'test f1': f1_score(y_test, y_pred),
})
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_smote, y_train_smote)
y_pred = knn.predict(X_test_preproced)
y_pred_train = knn.predict(X_train_smote)

print('train')
print(accuracy_score(y_train_smote, y_pred_train))
print(precision_score(y_train_smote, y_pred_train))
print(recall_score(y_train_smote, y_pred_train))
print(f1_score(y_train_smote, y_pred_train))
print(confusion_matrix(y_train_smote, y_pred_train))
print(classification_report(y_train_smote, y_pred_train))

print("-------test ----")
print(accuracy_score(y_test, y_pred))
print(precision_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(f1_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

result.append({
    'model':'knn',
    'train accuracy':accuracy_score(y_train_smote, y_pred_train),
    'test accuracy':accuracy_score(y_test, y_pred),
    'train precision': precision_score(y_train_smote, y_pred_train),
    'test precision': precision_score(y_test, y_pred),
    'train recall': recall_score(y_train_smote, y_pred_train),
    'test recall': recall_score(y_test, y_pred),
    'train f1 ':f1_score(y_train_smote, y_pred_train),
    'test f1': f1_score(y_test, y_pred),
})
```
