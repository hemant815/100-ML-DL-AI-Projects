```python
import pandas as pd
```

```python
df = pd.read_csv('weather_prediction_dataset.csv')
```

```python
df = df.drop_duplicates()
```

```python
df.isnull().sum().sum()
```

```python
measurements = ["temp_mean", "humidity", "pressure", "precipitation"]

cities = sorted({
    col.replace(f"_{measurement}", "")
    for col in df.columns
    for measurement in measurements
    if col.endswith(f"_{measurement}")
})

print("Number of cities:", len(cities))
print(cities)
```

```python
humidity_cols = [c for c in df.columns if "humidity" in c.lower()]
pressure_cols = [c for c in df.columns if "pressure" in c.lower()]
precip_cols = [c for c in df.columns if "precipitation" in c.lower()]
temp_cols = [c for c in df.columns if "temp_mean" in c.lower()]

print("Humidity:", len(humidity_cols))
print("Pressure:", len(pressure_cols))
print("Precipitation:", len(precip_cols))
print("Temperature:", len(temp_cols))
```

```python
df[humidity_cols] = df[humidity_cols].clip(0, 1)
df[pressure_cols] = df[pressure_cols].clip(0.8, 1.2)
df[precip_cols] = df[precip_cols].clip(lower=0)
```

```python
import warnings


warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', None)
```

```python
df['DATE'] = pd.to_datetime(df['DATE'], format='%Y%m%d')
df['month'] =  pd.to_datetime(df['DATE']).dt.month
```

```python
df["season"] = df["month"].map({
    12: "Winter", 1: "Winter", 2: "Winter",
    3: "Spring", 4: "Spring", 5: "Spring",
    6: "Summer", 7: "Summer", 8: "Summer",
    9: "Autumn", 10: "Autumn", 11: "Autumn"
})
```

```python
classification = pd.DataFrame(index=df.index)

classification["temperature"] = df[temp_cols].mean(axis=1)
classification["humidity"] = df[humidity_cols].mean(axis=1)
classification["pressure"] = df[pressure_cols].mean(axis=1)
classification["precipitation"] = df[precip_cols].mean(axis=1)
classification['season'] = df['season']

cloud_cols = [c for c in df.columns if "cloud_cover" in c.lower()]
sunshine_cols = [c for c in df.columns if "sunshine" in c.lower()]
wind_speed_cols = [c for c in df.columns if "wind_speed" in c.lower()]
radiation_cols = [c for c in df.columns if "global_radiation" in c.lower()]

classification["cloud_cover"] = df[cloud_cols].mean(axis=1)
classification["sunshine"] = df[sunshine_cols].mean(axis=1)
classification["wind_speed"] = df[wind_speed_cols].mean(axis=1)
classification["global_radiation"] = df[radiation_cols].mean(axis=1)
```

```python
classification.head()
```

```python
classification["rain_binary"] = (
    classification["precipitation"] > 0
).astype(int)
```

```python
classification.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14,10))
sns.heatmap(classification.corr(numeric_only=True),fmt='.2f', annot=True, cmap='coolwarm', cbar=True, linewidths=0.5)
plt.tight_layout()
plt.show()
```

```python
classification.head()
```

```python
classification['season'].value_counts()
```

```python
# One-hot encode season
classification = pd.get_dummies(
    classification,
    columns=["season"],
    drop_first=True,
    dtype=int
)
```

```python
classification.head()
```

```python
X = classification.drop(columns=['rain_binary', 'precipitation'])
y = classification['rain_binary']
```

```python
print(y.value_counts())
print(y.value_counts(normalize=True) * 100)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
X_train
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
X_train_scale = ss.fit_transform(X_train)
X_test_scale = ss.transform(X_test)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_scale,
    y_train
)
```

```python
print(y.value_counts())
print(y_train_smote.value_counts(normalize=True) * 100)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score

lr = LogisticRegression()
lr.fit(X_train_smote, y_train_smote)
y_pred = lr.predict(X_test_scale)
y_pred_train = lr.predict(X_train_smote)

print('train')
print(accuracy_score(y_train_smote, y_pred_train))
print(precision_score(y_train_smote, y_pred_train, average='weighted'))
print(recall_score(y_train_smote, y_pred_train , average='weighted'))
print(f1_score(y_train_smote, y_pred_train, average='weighted'))

print('test')
print(accuracy_score(y_test, y_pred))
print(precision_score(y_test, y_pred, average='weighted'))
print(recall_score(y_test, y_pred, average='weighted'))
print(f1_score(y_test, y_pred, average='weighted'))
```

```python
from sklearn.naive_bayes import GaussianNB


naive = GaussianNB()
naive.fit(X_train_smote, y_train_smote)
y_pred = naive.predict(X_test_scale)
y_pred_train = naive.predict(X_train_smote)

print('train')
print(accuracy_score(y_train_smote, y_pred_train))
print(precision_score(y_train_smote, y_pred_train, average='weighted'))
print(recall_score(y_train_smote, y_pred_train , average='weighted'))
print(f1_score(y_train_smote, y_pred_train, average='weighted'))

print('test')
print(accuracy_score(y_test, y_pred))
print(precision_score(y_test, y_pred, average='weighted'))
print(recall_score(y_test, y_pred, average='weighted'))
print(f1_score(y_test, y_pred, average='weighted'))
```

```python
from sklearn.ensemble import RandomForestClassifier


rf = RandomForestClassifier(n_estimators=101, max_depth=3)
rf.fit(X_train_smote, y_train_smote)
y_pred = rf.predict(X_test_scale)
y_pred_train = rf.predict(X_train_smote)

print('train')
print(accuracy_score(y_train_smote, y_pred_train))
print(precision_score(y_train_smote, y_pred_train, average='weighted'))
print(recall_score(y_train_smote, y_pred_train , average='weighted'))
print(f1_score(y_train_smote, y_pred_train, average='weighted'))

print('test')
print(accuracy_score(y_test, y_pred))
print(precision_score(y_test, y_pred, average='weighted'))
print(recall_score(y_test, y_pred, average='weighted'))
print(f1_score(y_test, y_pred, average='weighted'))
```
