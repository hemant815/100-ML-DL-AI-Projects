```python
X = df.drop('Class',axis=1)
y = df['Class']
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42, stratify=y_train)
```

```python
from sklearn.preprocessing import StandardScaler

scale = StandardScaler()
X_train_scale = scale.fit_transform(X_train)
X_val_scale = scale.transform(X_val)
X_test_scale = scale.transform(X_test)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train_scale, y_train)
print(y_train_smote.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, classification_report, recall_score,f1_score


lr = LogisticRegression(max_iter=1000)
lr.fit(X_train_smote, y_train_smote)

y_val_prob = lr.predict_proba(X_val_scale)[:,1]
print(y_val_prob)

y_val_pred = lr.predict(X_val_scale)
```

```python
for threshold in [0.5, 0.6, 0.7, 0.8, 0.9, 0.95, 0.99]:
    
    y_val_pred = (y_val_prob >= threshold).astype(int)
    
    precision = precision_score(y_val, y_val_pred, zero_division=0)
    recall = recall_score(y_val, y_val_pred, zero_division=0)
    
    print(
        f"Threshold={threshold} | "
        f"Precision={precision:.3f} | "
        f"Recall={recall:.3f}"
    )
```

```python
threshold = 0.7
y_test_prob = lr.predict_proba(X_test_scale)[:, 1]

y_test_pred = (y_test_prob >= threshold).astype(int)

print(accuracy_score(y_test, y_test_pred))
print(precision_score(y_test, y_test_pred))
print(confusion_matrix(y_test, y_test_pred))
print(classification_report(y_test, y_test_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier


rf = RandomForestClassifier(n_estimators=101)
rf.fit(X_train_smote, y_train_smote)

y_pred = rf.predict(X_test_scale)
y_pred_train = rf.predict(X_train_smote)

y_val_pro = rf.predict_proba(X_val_scale)[:,1]

y_val_pred = rf.predict(X_val_scale)
```

```python
for threshold in [0.5, 0.6, 0.7, 0.8, 0.9, 0.95, 0.99]:

    y_val_pred = (y_val_prob >= threshold).astype(int)

    precision = precision_score(y_val, y_val_pred)
    recall = recall_score(y_val, y_val_pred)
    f1 = f1_score(y_val, y_val_pred)

    print(
        f"Threshold={threshold} | "
        f"Precision={precision:.3f} | "
        f"Recall={recall:.3f} | "
        f"F1={f1:.3f}"
    )
```

```python
threshold = 0.99

y_test_prob = rf.predict_proba(X_test_scale)[:, 1]

y_test_pred = (y_test_prob >= threshold).astype(int)

from sklearn.metrics import (
    classification_report,
    confusion_matrix
)

print(precision_score(y_test, y_test_pred))
print(confusion_matrix(y_test, y_test_pred))
print(classification_report(y_test, y_test_pred))
```
