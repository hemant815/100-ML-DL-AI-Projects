```python
import pandas as pd
```

```python
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.xls')
```

```python
df.shape
```

```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
df = df.drop('customerID',axis=1)
```

```python
df.head()
```

```python
df.info()
```

```python
import numpy as np

df.replace(' ', np.nan, inplace=True)
df = df.dropna()

df['TotalCharges'] = df['TotalCharges'].astype(float)
```

```python
df.shape
```

```python
df.info()
```

```python
df.head()
```

```python
from sklearn.preprocessing import OrdinalEncoder, StandardScaler

ordinal_encode = OrdinalEncoder(dtype=np.int64)

df['Contract'] = ordinal_encode.fit_transform(df[['Contract']])
```

```python
df.head()
```

```python
df['Churn'] = df['Churn'].map({'Yes':1, 'No':0})
```

```python
df.info()
```

```python
df['tenure'] = df['tenure'].astype(float)
```

```python
df.info()
```

```python
X = df.drop('Churn',axis=1)
y = df['Churn']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

num_cols = X.select_dtypes(include='float').columns
cat_cols = X.select_dtypes(include='object').columns


preproccesor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
])
```

```python
X_train = preproccesor.fit_transform(X_train)
X_test = preproccesor.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, recall_score, confusion_matrix, classification_report


lr = LogisticRegression(class_weight='balanced')
lr.fit(X_train, y_train)

y_pred = lr.predict(X_test)
y_pred_train = lr.predict(X_train)
result = []

print('train')
print(accuracy_score(y_train, y_pred_train))
print(recall_score(y_train, y_pred_train))
print(confusion_matrix(y_train, y_pred_train))
print(classification_report(y_train, y_pred_train))

print("*"*20)
print('test')
print(accuracy_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

result.append({
    'model':'logistic',
    'train accuracy':accuracy_score(y_train, y_pred_train),
    'test accuracy': accuracy_score(y_test, y_pred),
    'train recall': recall_score(y_train, y_pred_train),
    'test recall' : recall_score(y_test, y_pred)
})
```

```python
from sklearn.naive_bayes import GaussianNB


naive = GaussianNB()
naive.fit(X_train, y_train)

y_pred = naive.predict(X_test)
y_pred_train = naive.predict(X_train)


print('train')
print(accuracy_score(y_train, y_pred_train))
print(recall_score(y_train, y_pred_train))
print(confusion_matrix(y_train, y_pred_train))
print(classification_report(y_train, y_pred_train))

print("*"*20)
print('test')
print(accuracy_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

result.append({
    'model':'naive',
    'train accuracy':accuracy_score(y_train, y_pred_train),
    'test accuracy': accuracy_score(y_test, y_pred),
    'train recall': recall_score(y_train, y_pred_train),
    'test recall' : recall_score(y_test, y_pred)
})
```

```python
from sklearn.neighbors import KNeighborsClassifier


knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
y_pred_train = knn.predict(X_train)


print('train')
print(accuracy_score(y_train, y_pred_train))
print(recall_score(y_train, y_pred_train))
print(confusion_matrix(y_train, y_pred_train))
print(classification_report(y_train, y_pred_train))

print("*"*20)
print('test')
print(accuracy_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

result.append({
    'model':'knn',
    'train accuracy':accuracy_score(y_train, y_pred_train),
    'test accuracy': accuracy_score(y_test, y_pred),
    'train recall': recall_score(y_train, y_pred_train),
    'test recall' : recall_score(y_test, y_pred)
})
```

```python
from sklearn.svm import SVC


svc = SVC(class_weight='balanced')
svc.fit(X_train, y_train)

y_pred = svc.predict(X_test)
y_pred_train = svc.predict(X_train)


print('train')
print(accuracy_score(y_train, y_pred_train))
print(recall_score(y_train, y_pred_train))
print(confusion_matrix(y_train, y_pred_train))
print(classification_report(y_train, y_pred_train))

print("*"*20)
print('test')
print(accuracy_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

result.append({
    'model':'svc',
    'train accuracy':accuracy_score(y_train, y_pred_train),
    'test accuracy': accuracy_score(y_test, y_pred),
    'train recall': recall_score(y_train, y_pred_train),
    'test recall' : recall_score(y_test, y_pred)
})
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=101, max_depth=3,class_weight='balanced')
rf.fit(X_train, y_train)
y_pred = rf.predict(X_test)
y_pred_train = rf.predict(X_train)

print('train')
print(accuracy_score(y_train, y_pred_train))
print(recall_score(y_train, y_pred_train))
print(confusion_matrix(y_train, y_pred_train))
print(classification_report(y_train, y_pred_train))

print("*"*20)
print('test')
print(accuracy_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))

result.append({
    'model':'random forest',
    'train accuracy':accuracy_score(y_train, y_pred_train),
    'test accuracy': accuracy_score(y_test, y_pred),
    'train recall': recall_score(y_train, y_pred_train),
    'test recall' : recall_score(y_test, y_pred)
})
```
